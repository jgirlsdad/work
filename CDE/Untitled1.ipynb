```python
import requests

# Replace these with the actual domain and dataset identifier
domain = "data.colorado.gov"
dataset_identifier = "7jm9-f28m"

# Construct the URL
url = f"https://{domain}/resource/{dataset_identifier}.json"
print(url)
# Query parameters
params = {
    "$where": "date>'2025-01-01T00:00:00'"
}

# Make the request
response = requests.get(url, params=params)

# Check if the request was successful
if response.status_code == 200:
    data = response.json()
    # Process the data
    for record in data:
        print(record)
else:
    print(f"Error: {response.status_code}")
```
