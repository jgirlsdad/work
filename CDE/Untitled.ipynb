```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', None)
```

```python
incidents = pd.read_csv("incidents-2017.csv")
```

```python
incidents.shape
```

```python
incidents.head()
```

```python
a = incidents["OFFENSE_CATEGORY_NAME"].value_counts()
b = pd.DataFrame(a)
print(b.shape)
b.to_html("incidetTypes.html")
```

```python
a = incidents.groupby(["UCR_AGENCY_NAME","AGENCY_TYPE_NAME"],as_index=False).count()
```

```python
a.shape
```

```python
a["AGENCY_TYPE_NAME"].value_counts().plot.bar("AGENCY_TYPE_NAME",title="Count of Police Agencies by Type in Colorado NIBRS Data")
plt.xlabel("Agency Type")
plt.ylabel("Count")

plt.savefig("AgencyType.png")
```

```python
incidents.sort_values("INCIDENT_HOUR").head()
```

```python
incidents["INCIDENT_HOUR"].value_counts().sort_index().plot.bar("INCIDENT_HOUR",title="Count of Incidents by Hour")

plt.xlabel("Hour of Incident")
plt.ylabel("Count")

plt.savefig("IncidentHour.png")
```
