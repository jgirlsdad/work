```python
import zipfile
import os
from pathlib import Path
```

```python

def extract_images_from_excel(excel_file, output_dir):
    """
    Extracts all images from a single Excel (.xlsx) file.
    Images are located in the xl/media directory inside the Zip archive.
    """
    excel_file = Path(excel_file)
    output_dir = Path(output_dir)

    # Create output directory for this workbook
    workbook_dir = output_dir / excel_file.stem
    workbook_dir.mkdir(parents=True, exist_ok=True)

    try:
        with zipfile.ZipFile(excel_file, 'r') as z:
            print("Starting")
            for file in z.namelist():
                print(file)
                # All embedded images live under xl/media/
                if file.startswith("xl/media/"):
                    filename = os.path.basename(file)
                    print(f"Found file in archive: {file}")
                    # Skip directories
                    if not filename:
                        continue

                    # Extract only image files
                    if filename.lower().endswith((
                        ".png", ".jpg", ".jpeg", ".gif",
                        ".bmp", ".tiff", ".svg"
                    )):
                        z.extract(file, workbook_dir)
                        print(f"[OK] Extracted {filename} from {excel_file.name}")
    except zipfile.BadZipFile:
        print(f"[ERROR] {excel_file.name} is not a valid .xlsx file")

def batch_extract(dir,excel_files, output_root="excel_images"):
    """
    Processes all .xlsx files in a folder.
    """
    #folder = Path(folder_with_excels)
    output_root = Path(output_root)

    output_root.mkdir(exist_ok=True)

    #excel_files = list(folder.glob("*.xlsx"))

    if not excel_files:
        print("No Excel files found.")
        return

    for f in excel_files:
        print(f"\n=== Processing {f} ===")
        extract_images_from_excel(f"{dir}/{f}", output_root)

    print("\nDONE. All images extracted.")

# ------------------------
# Run batch extraction
# ------------------------

# Example:
# batch_extract("C:/Users/me/Documents/excel_reports")
file="PPFS_2026 Ch 2 figures 2 thru 15.xlsx"
if __name__ == "__main__":
    batch_extract("OneDrive_1_10-30-2025",[file])
```
