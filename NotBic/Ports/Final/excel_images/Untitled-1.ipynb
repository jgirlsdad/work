```python
# extract_excel_images_optionD.py

import sys
from pathlib import Path
import win32com.client as win32

# Excel MsoShapeType Constants (we only export safe ones)
EXPORTABLE_TYPES = {
    4,   # msoPicture
    13,  # msoChart
    14,  # msoSmartArt (exportable only sometimes)
}

# Types we MUST skip (always crash)
SKIP_TYPES = {
    1,   # msoGroup
    5,   # msoEmbeddedOLEObject
    7,   # msoOLEControlObject
    12,  # msoFormControl
    24,  # msoShapeTypeMixed
    26,  # msoDiagram
}


def extract_from_excel(xlsx_path, out_dir):
    xlsx_path = Path(xlsx_path).resolve()
    out_dir = Path(out_dir).resolve()
    out_dir.mkdir(parents=True, exist_ok=True)

    excel = win32.Dispatch("Excel.Application")
    excel.Visible = False
    excel.DisplayAlerts = False

    print(f"\nOpening Excel workbook:\n  {xlsx_path}")
    wb = excel.Workbooks.Open(str(xlsx_path))

    img_index = 1

    for sheet in wb.Worksheets:
        print(f"\n=== SHEET: {sheet.Name} ===")

        # -----------------------------------
        # 1. Extract ChartObjects (Excel charts)
        # -----------------------------------
        try:
            for chart_obj in sheet.ChartObjects():
                out_file = out_dir / f"{sheet.Name}_chart_{img_index}.png"
                chart = chart_obj.Chart
                chart.Export(str(out_file))
                print(f"Exported ChartObject → {out_file}")
                img_index += 1
        except Exception as e:
            print(f"Error exporting ChartObjects in sheet {sheet.Name}: {e}")

        # -----------------------------------
        # 2. Extract Shapes (pictures, SmartArt, etc.)
        # -----------------------------------
        for shape in sheet.Shapes:
            shape_type = shape.Type

            # skip known non-exportable shapes
            if shape_type in SKIP_TYPES:
                continue

            if shape_type not in EXPORTABLE_TYPES:
                continue

            try:
                out_file = out_dir / f"{sheet.Name}_shape_{img_index}.png"
                
                shape.Export(str(out_file))
                print(f"Exported Shape (type={shape_type}) → {out_file}")
                img_index += 1

            except Exception as e:
                # Some SmartArt may still fail — skip gracefully
                print(f"Skipped shape type={shape_type} due to error: {e}")
                continue

    wb.Close(SaveChanges=False)
    excel.Quit()

    print("\nExtraction COMPLETE.")
    print(f"Images saved to: {out_dir}")


if __name__ == "__main__":
    

    xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFS_2026 Ch 2 figures 2 thru 15.xlsx"
  
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFS_2026 Ch 2 figure 16.xlsx"
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFS_2026 Ch. 3 graphs & tables.xlsx"
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFS Ch. 4 Top 25 Port Bridges.xlsx"
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFS_2026 Ch 4 Chinese Trade.xlsx"
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFS2026_508_Workbook.xlsx"
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFSP 2026 strait-of-hormuz.xlsx"
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\PPFSP Annual Report 2026 Appendix B.xlsx"
  #  xlsx_path = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\OneDrive_1_12-2-2025\Vessel Berthing Time graphs PPFS 2026 report.xlsx"
  
    out_dir = r"\\wsl.localhost\Ubuntu-24.04\home\joe\work\NotBic\Ports\excel_images"

    extract_from_excel(xlsx_path, out_dir)
```
