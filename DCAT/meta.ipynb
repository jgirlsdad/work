```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Step 1: Load your metadata (title, keywords, desc)
df = pd.read_csv("datasets_metadata.csv")   # or extract from DCAT TTL
df["text"] = (
    df["Dataset Title"].fillna('') + " " +
    df["Category"].fillna('') + " " +
    df["Short Description"].fillna('')
)

# Step 2: TF-IDF vectorization
vectorizer = TfidfVectorizer(stop_words="english", max_features=5000)
X = vectorizer.fit_transform(df["text"])

# Step 3: Clustering (try ~10–20 clusters)
k = 12
km = KMeans(n_clusters=k, random_state=0)
df["theme_cluster"] = km.fit_predict(X)

# Step 4: Examine top words per cluster
for i in range(k):
    idx = (km.labels_ == i)
    terms = X[idx].mean(axis=0).A1
    top_terms = [vectorizer.get_feature_names_out()[j] for j in terms.argsort()[-8:][::-1]]
    print(f"Cluster {i}: {', '.join(top_terms)}")
```
