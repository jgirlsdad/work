```python
import pandas as pd
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', None)
import math
```

```python
dfC = pd.read_csv("https://data.colorado.gov/resource/349y-twqi.csv")
dfN = pd.read_csv("Professional_and_Occupational_License_Types_in_Colorado_20241224.csv")
```

```python
dfC.shape,dfN.shape
```

```python
dfN.columns
import math
```

```python
histC={}
for indx,row in dfC.iterrows():
    cat = row["licensecategory"].strip()
    pre = row["licenseprefix"]
    if isinstance(pre,str) or not math.isnan(pre):
        pre=pre.strip()
    if cat not in histC:
        histC[cat]={}
    if pre not in histC[cat]:
        histC[cat][pre]=0
    histC[cat][pre]+=1
    


histN={}
for indx,row in dfN.iterrows():
    cat = row["licenseCategory"]
    pre = row["licensePrefix"]
    if isinstance(pre,str) or not math.isnan(pre):
        pre=pre.strip()
    if cat not in histN:
        histN[cat]={}
    if pre not in histN[cat]:
        histN[cat][pre]=0
    histN[cat][pre]+=1
    
```

```python
for cat,dct in histC.items():
    for pre,count in dct.items():
        if count > 1:
            print(cat,pre,count)
```

```python
for cat,dct in histN.items():
    for pre,count in dct.items():
        if count > 1:
            print(cat,pre,count)
```

```python
for cat,dct in histN.items():
    for pre,cnt in dct.items():
        if cnt > 1:
            print("BAD COUNT IN NEW DATA",cnt,cat,pre)
        if cat not in histC:
            print(f"NEW CAT IN NEW DATA {len(cat)} :{cat}: {pre}")
        elif pre not in histC[cat]:
            print(f"NEW TOPIC IN NEW DATA for CATEGORY {cat}: {pre}")
```

```python
for cat,dct in histC.items():
    for pre,cnt in dct.items():
        if cnt > 1:
            print("BAD COUNT IN CIM DATA",cnt,cat,pre)
        if cat not in histN:
            print(f"CIM CAT NOT IN NEW DATA {len(cat)}  {cat} {pre}")
        elif pre not in histN[cat]:
            print(f"CIM TOPIC IN NEW DATA for CATEGORY {cat}: {pre}")
```
