```python
def getData(file):
    fin=open(file)
    lines=fin.readlines()
    return lines

current=getData("bd.data.0.Current")
allItems=getData("bd.data.1.AllItems")
```

## Functions

```python
def colsToDict(df,key,val,fmt={}):
    return df.set_index(key)[val].to_dict()

def countDict(key,dct): 
    if key in dct:
        dct[key]+=1
    else:
        dct[key]=1

def decodeBDSeries(series):
    
    if series[0:2] == "BD":
       srs=series[2:]
    else:
       srs=series

    stats=srs[0:1]
    msa=srs[1:6]
    state=srs[6:8]
    county=srs[8:11]
    inds=srs[11:17]
    unit=srs[17:18]
    data=srs[18:19]
    rate=srs[23:24]
    period=srs[24:25]
       
    return stats,msa,state,county,inds,unit,data,rate,period

def createInvt(year,mon,invt):
    if year not in invt:
        invt[year]={}
    if mon not in invt[year]:
        invt[year][mon]=0
    invt[year][mon]+=1
    
```

```python
def processState(records,stateSelect="08"):
    histCols={}
    states={}
    counties={}
    footnotes={}
    invt={}
    histInds={}
    histUnit={}
    histData={}
    histRate={}
    histPeriod={}
    histSeries={}

    print("GETTING ",stateSelect)
    for mm,line in enumerate(records[1:]):
        spl=line.split("\t")
        nn=len(spl)
        year = int(spl[1])
        qtr = spl[2]
        foot = spl[3]
      
        stats,msa,state,county,inds,unit,data,rate,period=decodeBDSeries(spl[0])
        if state == stateSelect:
            countDict(nn,histCols)
            countDict(state,states) 
            countDict(county,counties)
            countDict(foot,footnotes)
            countDict(inds,histInds)          
            countDict(unit,histUnit)          
            countDict(data,histData)          
            countDict(rate,histRate)          
            countDict(period,histPeriod)          
            countDict(spl[0],histSeries)          
            
            createInvt(year,qtr,invt)
            
    print("MM ",mm)
    print("STATE ",state,year,qtr)
    return histCols,states,counties,footnotes,invt,histInds,histUnit,histData,histRate,histPeriod,histSeries

histColsAllCo,statesAllCo,countiesAllCo,footnotesAllCo,invtAllCo,histIndsCo,histUnitCo,histDataCo,histRateCo,histPeriodCo,histSeries = processState(allItems,"08")
```
