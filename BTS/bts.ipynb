```python
df1 = dfOldNo2016NoExtra
df2 = dfNewNoDwellNo2021
stats = {}
newPorts = {}

bad = {}
for port in df1["Port_Name"].unique().tolist():
    tmp1 = df1.loc[df1["Port_Name"] == port]
    tmp2 = df2.loc[df2["Port_Name"] == port]
    stats[port] = {}
    
#    for year in tmp1["Reporting Year"].unique().tolist():
    for year in [2017,2018,2019,2020]:
        
        tO = tmp1.loc[tmp1["Reporting Year"].astype(int) == year]
        tN = tmp2.loc[tmp2["Reporting Year"].astype(int) == year]
        if tO.shape[0] > 0:
            print(port,year,tO.shape[0],tN.shape[0])
            tO.reset_index(inplace=True,drop=True)
            tN.reset_index(inplace=True,drop=True)

    #         print(tO.columns)
    #         print(tN.columns)
            tN["Reporting Year"] = tN["Reporting Year"].astype(int)
            col='Percent Change'
            tN[col]= tN[col].str.replace(",","")
            tN[col] = tN[col].astype(float)
            tO[col] = tO[col].apply(lambda x: '{:,.2f}'.format(x))
            tN[col] = tN[col].apply(lambda x: '{:,.2f}'.format(x))
            col='Volume'
            tN[col] = tN[col].str.replace(",","")
            tN[col] = tN[col].astype(float)
    #         print("-------------------\nOLD")
    #         print(tO)
    #         print("-------------------\nNEW")

    #         print(tN)

            # tO[col] = tO[col].apply(lambda x: '{:,.2f}'.format(x))
            # tN[col] = tN[col].apply(lambda x: '{:,.2f}'.format(x))
            if tN.shape[0] > 0:
                a = tO.compare(tN)
                if len(a) > 0:
                    print(tO.compare(tN))
                    stats[port][year]="F"
                    if port not in bad:
                        bad[port] = {}
                    if year not in bad[port]:
                        bad[port][year]=[]
                    bad[port][year].append(a)
                else:
                    stats[port][year]="P"
            else:
                stats[port][year]="N"
                if port not in newPorts:
                    newPorts[port]={}
                newPorts[port][year]=1
        else:
                if tN.shape[0] == 0:
                    stats[port][year]=" "
                else:
                    stats[port][year]="O"
            
                
        
```
