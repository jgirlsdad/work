```python
file="/home/joe/bic_etl/cdos/business/business/data_source/corpmstr.txt"
fin=open(file,"rb")
detector = chardet.universaldetector.UniversalDetector()
nlines=0
for line in fin: 
    nlines+=1
    if nlines%500000 == 0:
        print(nlines)
    detector.feed(line) 
    if detector.done: 
        break
    if nlines > 1000000:
        break
detector.close() 
print(detector.result)
```

```python
file="/home/joe/bic_etl/cdos/business/business/data_source/corpmstr.txt"
fin=open(file,"r",encoding='ISO-8859-1')
hist={}
nlines=0
for line in fin: 
    nlines+=1
    if nlines%500000 == 0:
        print(nlines)
    for c in line:
        nc = ord(c)
        if nc not in hist:
            hist[nc]=0
        hist[nc]+=1
```
