```python
#fin = open("jss","r")
fin = open("cdos_jss","r")

lines = fin.readlines()
targets = {}
for line in lines:
    nst = line.find("(")
    ned = line[nst:].find(")")
    if (nst > -1):
   #   print(line)
      target = line[nst+1:nst+ned]
      if target not in targets:
         targets[target]=0

      targets[target]+=1
    #   print(target)
    #   print("-------------------------------")
```

```python
hit=0
miss=0
skipped = {}
fhit = open("libsToLoad","w")
finv = open("libsToLoadInv","w")

fskp = open("libsToSkip","w")

for key,val in sorted(targets.items(), key=lambda x:x[1]):
    if key.find("/") == -1:
      key = key.replace("'","")
      key = key.replace('"','')
      print(key)
      finv.write(f"{val:4d} - {key}\n")
      fhit.write(f"RUN npm install {key} -g\n")
        
      hit+=1
    else:
      miss+=1
      skipped[key] = val
      fskp.write(f"{val:4d} - {key}\n")

fhit.close()
finv.close()
fskp.close()

print(f"{hit}  Found,  skipped: {miss}")
   
```
