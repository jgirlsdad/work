```python
import pandas as pd
import os
```

```python
#Ideal UX: import module into a .py or .pynb in scripts and output a csv to qa_reports folder.
#Module would live with the qa scripts. Probably under the CDOS suite. 
#method / function for individual dataset analysis

def source_path(dataset_name):
    '''takes in dataset name and returns its path'''
    
    return f'../data_source/{dataset_name}'

def txt_to_list(broken_dataset) -> list:

    '''Reads tsv txt and returns dataset as a list of rows.'''

    rows_list = []
    
    with open(broken_dataset,'r',encoding = 'ISO-8859-1') as dataset:
        for line in dataset:
            rows_list.append(line.split('\t'))

    return rows_list

def extract_bad_records(processed_dataset):

    'takes in processed tsv txt file as a list'
    expected_row_length = len(processed_dataset[0]) #assumes first row is for header.
    print(f'Column names are:{processed_dataset[0]}') 
    
    bad_records = {}
    
    for index,line in enumerate(processed_dataset):
        if len(line) != expected_row_length:
            bad_records[index] = line
    return bad_records, expected_row_length

    #todo: length of row_list attribute for length of dataset capture when opening tsv. 

def bad_records_report(bad_records, dataset,expected_row_length):
    '''generate a csv report of the offending rows.'''
    
    report = pd.DataFrame(columns = ['Index_Number','Row_Contents', 'Bug_Comments'])
    report['Index_Number'] = bad_records.keys()
    report['Row_Contents'] = bad_records.values()
    report['Bug_Comments'] = f"row does not match expected {expected_row_length} values."
    os.makedirs('../qa_reports/', exist_ok = True)
    report.to_csv(f'../qa_reports/{dataset}_bugged_rows.csv', index = False, columns = ['Index_Number','Row_Contents', 'Bug_Comments'])
    print(f" {len(bad_records)} offending rows out of f'placeholder' were found and given in the dictionary below. The dictionary marks row index:[row's contents].\n")

    return report
```

```python
file_names = {
    "Charity Extension Requests" : "char_orgs_ext.tsv",
    "Paid Solicitors Disclosed on Charity Registration Forms in Colorado": "char_orgs_sol.tsv",
    "Other Names a Registered Entity Uses to Solicit Contributions":"doing_bus_as.tsv",
    "Persons Associated with Charitable Organizations, Paid Solicitors, and Professional Fundraising Consultants in Colorado":"persons_entity.tsv",
    "Paid Solicitor Solicitation Notices in Colorado":"sol_ntcs.tsv",
    "Registration for Charities, Paid Solicitors, Professional Fundraising Consultants, and for-profit Public Benefit Corporations in Colorado":"reg_finan.tsv",
} #dataset name : dataset filepath
 
```

```python
failing_datasets =  list(file_names.values())#data_source names of datasets.
```

```python
for dataset in failing_datasets:
    filename_path = source_path(dataset)
    rows_list = txt_to_list(filename_path)
    expected_row_length = len(rows_list[0])
    bad_records, expected_row_length = extract_bad_records(rows_list)
    report = bad_records_report(bad_records, dataset, expected_row_length) #fix this bad_records[1] into an expected row length attribute per dataset instance when standadrizing.
```
