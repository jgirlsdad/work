```python
import sqlite3
import pandas as pd
connr = sqlite3.connect('Data/DB/era5DataMeans.db')
tmp = pd.read_sql("select * from MEANS_TTdRHVPD",connr)

dates = tmp["Yrmo"].values.tolist()
invt={}
for yrmo in dates:
    yr=int(yrmo[:4])
    mo=int(yrmo[4:6])
    if yr not in invt:
        invt[yr]={}
    if mo not in invt[yr]:
       invt[yr][mo]=0

    invt[yr][mo]+=1
 

for year in sorted(invt.keys()):
    print(f"{int(year):4d}",end="")
    for month in range(1,13):
        if month in invt[year]:
            cnt=invt[year][month]
        else:
            cnt=0
        print(f"{cnt:5d}",end="")
    print("")
```
