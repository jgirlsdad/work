```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df= pd.read_csv("monthly_mae_adaptive.csv")
```

```python
df.head()
```

```python
for month in range(1,13):
    mdf=df[df['month']==month]
    mdf.plot(x="horizon",y="mae")
```

```python
df_pivot = df.pivot(index='horizon', columns='month', values='mae')
df_pivot.plot(figsize=(10, 6), marker='o')
```
