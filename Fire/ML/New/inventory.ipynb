## Daily Data

```python

import sqlite3, pandas as pd
from pathlib import Path
TARGET_VAR = '2d'
sqlite_path = Path(f"/home/joe/work/Fire/ML/Data/DB/era5_daily_2982_{TARGET_VAR}.sqlite")
table_name = "daily_data"
with sqlite3.connect(sqlite_path) as conn:
    sample = pd.read_sql_query(f"SELECT datetime FROM {table_name}", conn)

# with sqlite3.connect(sqlite_path) as conn:
    
#     for i, row in latlons.iterrows():
        
#         pid = row['point']
#         sample = pd.read_sql_query(f"SELECT * FROM {table_name} where point_id = {pid}", conn)
```

```python
sample['yrmo'] = sample['datetime'].str.slice(0, 6).astype(int)
a=sample.groupby('yrmo',as_index=False).size()

hist={}
for _, row in a.iterrows():
    yr = row['yrmo'] // 100
    mo = row['yrmo'] % 100
    if yr not in hist:
        hist[yr] = {}
    hist[yr][mo] = row['size']

    
for yr in range(1990,2027):
    print(f"{yr} ",end='')
    if yr not in hist:
        hist[yr] = {mo: 0 for mo in range(1, 13)}
    for mo in range(1, 13):
        if mo in hist[yr]:
            cnt=hist[yr][mo]
        else:
            cnt=0
        print(f" {cnt:06d}", end='')
    print()
```
