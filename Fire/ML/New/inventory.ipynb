```python
b=sample[["0","3","6","9","12","15","18","21"]].sum(axis=1)
```

```python
b.head()
```

```python
import numpy as np
import pandas as pd
from sklearn.mixture import GaussianMixture
from scipy.optimize import brentq
from scipy.stats import norm

# precip: 1D array-like of daily precip values (>=0)
x = np.asarray(b.values, dtype=float)
x = x[np.isfinite(x)]
x = x[x >= 0]

# keep strictly positive values for fitting
xp = x[x > 0]

# optional: trim extreme storms so they don't dominate fit
hi = np.quantile(xp, 0.995)
xp_fit = xp[xp <= hi]

# log transform
y = np.log10(xp_fit).reshape(-1, 1)

gmm = GaussianMixture(n_components=2, random_state=0)
gmm.fit(y)

w = gmm.weights_
m = gmm.means_.ravel()
s = np.sqrt(gmm.covariances_.ravel())

# identify lower-mean component as "dry/noise", higher as "wet"
i_dry, i_wet = np.argsort(m)

def f(z):
    dry = w[i_dry] * norm.pdf(z, m[i_dry], s[i_dry])
    wet = w[i_wet] * norm.pdf(z, m[i_wet], s[i_wet])
    return dry - wet

# solve in log space between the two means
z_cut = brentq(f, m[i_dry], m[i_wet])
cutoff = 10 ** z_cut

print("Estimated cutoff:", cutoff)

# apply cutoff
x_clean = np.where(x < cutoff, 0.0, x)
```
