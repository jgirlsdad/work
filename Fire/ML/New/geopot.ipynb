```python
import xarray as xr
import pandas as pd
import numpy as np

# open dataset
ds = xr.open_dataset("Data/geopotential-42x71.nc")

# convert geopotential to elevation (meters)
elevation = ds["z"] / 9.80665

# get coordinate arrays
lats = ds["latitude"].values
lons = ds["longitude"].values

# create full grid
lon_grid, lat_grid = np.meshgrid(lons, lats)

# flatten everything
df = pd.DataFrame({
    "latitude": lat_grid.ravel(),
    "longitude": lon_grid.ravel(),
    "elevation_m": elevation.values.squeeze().ravel()
})
df['point'] = df.index #  Add a unique point ID starting from 0
print(df.head())
print(f"\nTotal points: {len(df)}")
df.to_csv("Data/latlon_elevation.csv", index=False)
```
