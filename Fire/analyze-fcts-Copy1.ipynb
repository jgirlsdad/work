```python
df["validDate"] = pd.to_datetime(df["validDate"])
df["forecastDate"] = pd.to_datetime(df["forecastDate"])
df["forecastDate"] = df["forecastDate"].dt.date
df["validDate"] = df["validDate"].dt.date
```

```python
vrs = ["p1","p2","p3","p4","p5","p6","p7","p8","p9","p10","p11","p12","p13","p14","p15","p16","p17","p18","p19","p20","p21","p22","p23","p24","p25","p26","p27","p28","p29","p30","p31","p32"]

nper=5
minDate = df.loc[(df["valueOfForecastTime"] == 0),"validDate"].min()
start = minDate+relativedelta(days=nper)
```

```python
hist={}
dfOut = pd.DataFrame(columns=["parameter","level","validDate","forecastDate","forecastPeriod","window","point",
                              "observedValue","forecastValue","difference",
                              "numberOfForecastValues","forecastStdDev"])
tyearS = 2018
tyearE = 2023
for tdate in df.loc[(df["valueOfForecastTime"] == 0) & (df["validDate"] >= start),"validDate"].sort_values():
    year = tdate.year
    month = tdate.month 
    day = tdate.day
    bin = round(day/10,0)
    if year < tyearS:
        continue
    if year > tyearE:
        break
    if bin > 2:
        bin=2
    print(tdate)
    for pt in vrs: 
        
        for nf in range(1,10):
            sdate = tdate - relativedelta(days=nper)
            edate = tdate + relativedelta(days=nper)
            fdate = tdate + relativedelta(months=nf)  #  forecast date
            fsdate = fdate - relativedelta(days=nper)  #  forecast date
            fedate = fdate + relativedelta(days=nper)  #  forecast date
            an = df.loc[(df["valueOfForecastTime"] == 0) & (df["validDate"] == fdate),["shortName","level",pt]]
            if an.shape[0] < 1:
                continue
            par = an["shortName"].values[0]
            level = an["level"].values[0]
            anl = an[pt].values[0]
            fcs = df.loc[(df["valueOfForecastTime"] == nf) & (df["forecastDate"] >= fsdate) & (df["forecastDate"] <= fedate),pt]
            fv = fcs.mean()
            std = fcs.std()
            
            diff = abs(anl- fv)
            tdate,pt,nf,fdate,fsdate,fedate,fv,fcs,diff
            tmp = {
                   "parameter":[par],
                       "level":[level],
                   "validDate":[tdate],
                "forecastDate":[fdate],
              "forecastPeriod":[nf],
                      "window":[nper],
                       "point":[pt],
               "observedValue":[anl],
               "forecastValue":[fv],
                  "difference":[diff],
      "numberOfForecastValues":[fcs.shape[0]],
              "forecastStdDev":[std]}
            tmp = pd.DataFrame(tmp)
            dfOut = pd.concat([dfOut,tmp])
            # if pt not in hist:
            #     hist[pt]={}
            # if month not in hist[pt]:
            #     hist[pt][month]={}
            # if bin not in hist[pt][month]:
            #     hist[pt][month][bin]={}
            #     hist[pt][month][bin]["sum"]=0
            #     hist[pt][month][bin]["count"]=0
                

            # hist[pt][month][bin]["sum"]+=abs(diff)
            # hist[pt][month][bin]["count"]+=abs(diff) 
            
    
 #           print(pt,tdate,sdate,edate,fdate,anl,fv,diff)
 
```
