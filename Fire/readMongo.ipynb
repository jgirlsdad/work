```python
def computeStats(row):
    test=row["y_test"]
    hat=row["y_hat"]
    testStd= statistics.stdev(test)
    hatStd= statistics.stdev(hat)
    dif = [abs(test[nn]-hat[nn]) for nn in range(len(test))]
    difStd=statistics.stdev(dif)
    row["y_test_std"],row["y_hat_std"],row["y_diff_std"] = testStd,hatStd,difStd
    return row

dfNew = df.apply(computeStats,axis=1)
    
```
