```python
import numpy as np

def rh_from_t_td(t, td, units="C"):
    """Compute relative humidity (%) from air temperature and dew point.

    Parameters
    ----------
    t : float or array-like
        Air temperature.
    td : float or array-like
        Dew point temperature (same units as t).
    units : str
        "C" for Celsius (default) or "K" for Kelvin.
    """
    t = np.asarray(t, dtype=float)
    td = np.asarray(td, dtype=float)

    if units.upper() == "K":
        t = t - 273.15
        td = td - 273.15
    elif units.upper() != "C":
        raise ValueError("units must be 'C' or 'K'")

    # Magnus formula (valid for typical atmospheric conditions)
    rh = 100.0 * np.exp((17.625 * td) / (243.04 + td)) / np.exp((17.625 * t) / (243.04 + t))
    return np.clip(rh, 0.0, 100.0)

# Example
print(rh_from_t_td(30.0, 20.0, units="C"))
```

```python

def tTd2RH(t,td):
#  t,td should be in C
    rh=  100 * (math.exp((17.625 * td) / (243.04 + td))) / (math.exp((17.625 * t) / (243.04 + t)))
    return rh

def vpd(T,RH):
#    T should in in C
 #   tt=T-273.15
    tt=T
    es =0.6108*math.exp((17.27*tt)/(tt+237.3))
    if RH>1:
       val=100
    else:
        val=1
    ea=RH/val*es
    return es-ea

def decodeGrib(grb,data):
    dataTmp=grb.data()
    nlons=29
    date=f'{grb["dataDate"]}{str(grb["validityTime"]).zfill(4)}'
    yrmo=f'{grb["year"]}{grb["month"]}'
#    print(date)
    if date not in data:
        data[date]={}
    tmp = np.empty((17,29))
    latslons= []
    data[date][grb['paramId']]=dataTmp[0]
    # for nrow,dat in enumerate(dataTmp[0]):
    #     # print(nrow,len(dat),data[0][nrow])
    #    ll=[]
    #    for ncol,val in enumerate(dat):
    #        val=dataTmp[0][nrow][ncol]
    #        pt=nlons*nrow+ncol
    #        var=grb['paramId']
    #        if pt not in data[date]:
    #           data[date][pt]={}
          
              
               
      # #     print(nrow,ncol,pt,dataTmp[2][nrow][ncol],dataTmp[1][nrow][ncol])
      #  latslons.append(ll)
    return data


def combineData(data):
    allData={}
    print("COMBINED DATA LEN ",len(data))
    for date in sorted(data.keys()):
     #   print(date)
        tDat=data[date][167]
        tDDat=data[date][168]
        yrmo=date[:6]
        for nrow in range(0,nlats):
            for ncol in range(0,nlons):
              T=tDat[nrow][ncol]-273.15
              Td=tDDat[nrow][ncol]-273.15
              RH=tTd2RH(T,Td)
              vp=  vpd(T,RH)
              pt=nlons*nrow+ncol
              # if pt > 493:
              #     print(pt,nrow,ncol)
                  
              if pt not in allData:
                  allData[pt]={}
              if yrmo not in allData[pt]:
                  allData[pt][yrmo] = {}
              if date not in allData[pt][yrmo]:
                  allData[pt][yrmo][date]={}
              allData[pt][yrmo][date]["T"]=round(T,2)
              allData[pt][yrmo][date]["Td"]=round(Td,2)
              allData[pt][yrmo][date]["RH"]=round(RH,2)
              allData[pt][yrmo][date]["vpd"]=round(vp,4)
    return allData


def computeMeans(allData):
    dates=[]
    temps=[]
    dews=[]
    rhs=[]
    vpds=[]
    pts=[]
    yrmos=[]
    lens=[]
    print("COMPUTE MEANS ALL DATA LEN ",len(allData))
    for pt,dct in allData.items():
        for yrmo,dct1 in dct.items():
            datesTmp=[]
            tempsTmp=[]
            dewsTmp=[]
            rhsTmp=[]
            vpdsTmp=[]
            for date,dct2 in dct1.items():
                datesTmp.append(date)
                tempsTmp.append(dct2["T"])
                dewsTmp.append(dct2["Td"])
                rhsTmp.append(dct2["RH"])
                vpdsTmp.append(dct2["vpd"])
            if len(tempsTmp) > 160:  # need at-least about 2/3 of a month of data to use  means
                tMean=sum(tempsTmp)/len(tempsTmp)
                tdMean=sum(dewsTmp)/len(dewsTmp)
                rhMean=sum(rhsTmp)/len(rhsTmp)
                vpdMean=sum(vpdsTmp)/len(vpdsTmp)
                pts.append(pt)
                yrmos.append(yrmo)
                temps.append(tMean)
                dews.append(tdMean)
                rhs.append(rhMean)
                vpds.append(vpdMean)
                lens.append(len(vpdsTmp))
    print("MEANS LENS ",len(yrmos),len(pts),len(temps),len(dews),len(rhs),len(vpds),len(lens))
    dfOut = pd.DataFrame({
        "Point":pts,
        "Yrmo":yrmos,
        "Length":lens,
        "T":temps,
        "Td":dews,
        "RH":rhs,
        "VPD":vpds
    })
    return dfOut
```
