```python
import pygrib
import numpy as np
import sqlite3
import pandas as pd
import math
import time,os
import matplotlib.pyplot as plt
# from mpl_toolkits.basemap import Basemap
import glob
```

```python
#fin=open("Data/list2")
#files=fin.readlines()
#files=glob.glob("Data/*.grib")
files=glob.glob("era5_downloads/*.grib")

print(files)
invt={}
invtVar={}
fout=open("files_summary3.txt","w")
for file in files:
    file=file.strip()
#    print(file)
   # grbs = pygrib.open(f"Data/{file}")
    grbs = pygrib.open(f"{file}")
    
    years={}
    months={}
    vars={}
#    print(len(grbs))
#    print(grbs[1].numberOfDataPoints)
    if grbs[1].numberOfDataPoints == 2982:
        for grb in grbs:
           yr=grb["year"]
           mo=grb["month"]
           dy=grb["day"]
           var=grb['paramId']
           name=grb['shortName']
           var=name
           if var not in vars:
               vars[var]=0
           vars[var]+=1
           if yr not in years:
               years[yr]=0
           if mo not in months:
               months[mo]=0
           months[mo]+=1
           years[yr]+=1
           if var not in invt:
               invt[var]={}
           if yr not in invt[var]:
               invt[var][yr]={}
           if mo not in invt[var][yr]:
               invt[var][yr][mo]=0
    
           invt[var][yr][mo]+=1
       
        # for yr in sorted(years.keys()):
        #     print(f"   {yr}  {years[yr]:7d}")
        # for mo in months:
        #     print(f"     {mo}  {months[mo]:5d}")
        fout.write(f"{file} {name} {vars.keys()}  {sorted(years.keys())}\n")
        print(f"{file} {name} {vars.keys()}  {sorted(years.keys())}\n")
        for yr in sorted(years.keys()):
            print(f"   {yr}  {years[yr]:7d}")
        print("\n")
        grbs.close()
fout.close()
print("DONE")
```

```python
annual={}
vars={}
for var,dct in invt.items():
    for year,months in sorted(dct.items()):
        totl=sum(months.values())
        if year not in annual:
            annual[year]={}
        if var not in annual[year]:
            annual[year][var]={}
        annual[year][var]=totl
        if var not in vars:
            vars[var]=0
        vars[var]+=totl

print(f"Year   ",end="")
for var in sorted(vars.keys()):
    print(f"{var:6s}",end="")
print("\n")
for year,dct in sorted(annual.items()):
    print(f"{year}",end="")
    for var in sorted(vars.keys()):
        if var in dct:
            totl=dct[var]
        else:
            totl=0
        print(f"{totl:6d}",end="")
    print("")
```
