# __Accessing Database Using SQLite__


## __Objective__
To demonstrate the connection of an SQLite database, creating a table, inserting, updating, and deleting records, followed by displaying the data using Pandas DataFrame, illustrating CRUD operations in SQLite with Python.

## __Steps to Perform__

1. Import SQLite3 and Pandas
2. Connect to or create the SQLite Database and Initialize a Cursor
3. Create the "students" Table if it Doesn't Exist
4. Insert Initial Data into the "students" Table
5. Read Data from "students" into a DataFrame and Print it
6. Update John's age in the Table, Read Updated Data into a DataFrame, and Print it
7. Delete Emma's Record from the Table, Read Updated Data into a DataFrame, and Print it

### Step 1: Import SQLite3 and Pandas

```python
import sqlite3
import pandas as pd
```

### Step 2: Connect to or Create the SQLite Database and Initialize a Cursor

```python
# Connect to the database or create if it doesn't exist
conn = sqlite3.connect('example.db')
# Create a cursor object to interact with the database
cursor = conn.cursor()
```

```python
type(cursor)
```

```python
dir(cursor)
```

### Step 3: Create the "students" Table if it Doesn't Exist

```python
# Create a "students" table if it doesn't exist
cursor.execute('''CREATE TABLE IF NOT EXISTS students
 (id INTEGER PRIMARY KEY,
 name TEXT NOT NULL,
 age INTEGER NOT NULL)''')
```

### Step 4: Insert Initial Data into the "students" Table

```python
# Insert data into the "students" table
cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ('John', 25))
cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ('Emma', 22))
```

### Step 5: Read Data from "students" into a DataFrame and Print it

```python
# Read data from the "students" table into a pandas DataFrame
df = pd.read_sql_query("SELECT * FROM students", conn)
# Print the DataFrame
print(df)
```

### Step 6: Update John's age in the Table, Read Updated Data into a DataFrame, and Print it

```python
# Update data in the "students" table
cursor.execute("UPDATE students SET age = ? WHERE name = ?", (23, 'John'))
df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)
```

### Step 7: Delete Emma's Record from the Table, Read Updated Data into a DataFrame, and Print it

```python
# Delete a student from the "students" table
cursor.execute("DELETE FROM students WHERE name = ?", ('Emma',))
df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)
```
