```python
a = ['A','B','C','A','B']
c = ['one','cat','mouse']

symbols = {}
for val in a:
    if val not in symbols:
        symbols[val] = {}
        symbols[val]['ID'] = c[0]
        symbols[val]['CLASS'] = c[1]        
        symbols[val]['rodent'] = c[2]        
```

```python
for symb in symbols.keys():
    print(symb)
    print(symbols[symb]['number'],symbols[symb]['animal'])
```
