```python
def getXrefs():
    '''Reads the Inventory google sheet and gets the datasets title and cross-refs it to the Socrata 4x4 id.  Also
    gets the fields by 4x4 dataset id and by the title'''
    scope = ['https://www.googleapis.com/auth/spreadsheets.readonly',
             "https://www.googleapis.com/auth/drive.file",
                  "https://www.googleapis.com/auth/drive"]

    creds = ServiceAccountCredentials.from_json_keyfile_name('../client_secret.json',
     scope)
    client = gspread.authorize(creds)

    gc = gspread.service_account("../client_secret.json")
    # for gg in gc.list_spreadsheet_files():
    #      print("GGGGG ",gg)
    # https://docs.google.com/spreadsheets/d/1WTaOglzbSsYiHhAGguGxHQXmAGmOhfFHkGkMLowxAOA/edit?usp=sharing
#    sheet = client.open('BIC Data Inventory and Metadata').worksheet(
#        'Maintenance_Framework')
    inv_sheet = client.open('BIC Data Inventory and Metadata').worksheet(
        'MetadataRepository')
    
    org_sheet = client.open('Copy of bic_metadata_updater').worksheet(
        'Sheet1')
    

    dfNew = pd.DataFrame(inv_sheet.get_all_records(head=3))
    dfOrg = pd.DataFrame(org_sheet.get_all_records(head=1))
    
    return dfNew,dfOrg
   
dfN,dfO = getXrefs()
```

```python
import requests,os,json,sys
from bs4 import BeautifulSoup
bic_etl_home = os.getenv('bic_etl_home')
sys.path.insert(0, os.path.join(bic_etl_home, 'general', 'scripts'))

with open(os.path.join(bic_etl_home, 'general', 'datasync', 'config.json')) as f:
    config = json.load(f)

socrata_password = config["password"]


base_url = 'https://data.colorado.gov/api/views/metadata/v1/'
username = "oit_datacoloradogov@state.co.us"
password = socrata_password
print(password)
# URL to the login page
login_url = 'https://data.colorado.gov/login'

# Start a session to persist cookies
session = requests.Session()

# Get the login page to retrieve any necessary hidden form fields
login_page = session.get(login_url)
soup = BeautifulSoup(login_page.content, 'html.parser')

# Find hidden form fields (if any)
hidden_inputs = soup.find_all('input', type='hidden')
form_data = {input.get('name'): input.get('value') for input in hidden_inputs}

# Add login credentials to the form data
form_data['username'] = username
form_data['password'] = password

# Submit the login form
response = session.post(login_url, data=form_data)

# Check if login was successful
if response.url == 'https://data.yoursocrataportal.com/dashboard':
    print('Login successful')
    # Now you can access other pages that require login
    activity_log_url = 'https://data.yoursocrataportal.com/resource/xxxx-xxxx'
    activity_log_page = session.get(activity_log_url)
    
    # Process the activity log page as needed
    soup = BeautifulSoup(activity_log_page.content, 'html.parser')
    print(soup.prettify())
else:
    print('Login failed')
```
