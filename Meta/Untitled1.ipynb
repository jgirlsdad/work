```python
import pandas as pd
from sodapy import Socrata
import pandas as pd
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', None)
```

```python
cim_url_query = 'data.colorado.gov'
datasets = None

with Socrata(cim_url_query, None) as client:
    datasets = client.datasets()
```

```python
for dat in datasets:
     if dat["resource"]["name"] == "Paid Solicitor Solicitation Notices in Colorado":
        break
```

```python
import re
titles=[]
hist={}
for dataset in datasets:
    own = dataset['owner']['display_name']    
    if own in hist:
        hist[own]+=1
    else:
        hist[own] = 1
    if own == "Colorado Information Marketplace":
        title = dataset['resource']['name']
        title=title.strip()
        n = re.search("map",title.lower())
        if n:
            titles.append(title)
    
```

```python
fout = open("tmp.txt","w")
for title in titles:
    n = title.lower().find("map")
    if n > 0:
        print(title)
        fout.write(f"{title}\n")
        
fout.close()
```
