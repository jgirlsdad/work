```python
def listDiff(a,b):
    ''' Compares 2 lists and returns the items unique to the first list, the item unique
        to the second list and items common between the 2 lists
        
        The return order is :   unique list a, unique list b, common both
        
        '''
    aa = [col.lower() for col in a.columns]
    bb = [col.lower() for col in b.columns]
    
    aSet = set(aa)
    bSet = set(bb)  
    
    adiff = list(aSet-bSet)
    bdiff = list(bSet-aSet) 
    inter = list(aSet.intersection(bSet))
    
    return adiff,bdiff,inter

a,b,c = listDiff(sOrigCols,sConvCols)
```
