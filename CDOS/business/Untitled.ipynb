```python
fin1 = open("/home/joe/bic_etl/cdos/business/business/data_source/business_entities.tsv","rb")

linesOrg = fin1.readlines()
fin1.close()

#fin2 = open("business_entities-new.csv")
fin2 = open("/home/joe/bic_etl/cdos/business/business/data_transformed/business_entities.tsv")
linesTrn = fin2.readlines()
fin2.close()

```

```python
linesTrn[1]
```

```python
histOrg={}
histTrn={}

for mm,line in enumerate(linesOrg[1:]):
    line2=linesTrn[mm+1]
  #  print(len(line),len(line2))
    if mm%100000 == 0:
        print("Line ",mm)
    for nn,c in enumerate(line[:-2]):
        c2 = line2[nn:nn+1]
        c2=ord(c2)
        if c2 not in histTrn:
            histTrn[c2]=0
        histTrn[c2]+=1
        
        if c not in histOrg:
            histOrg[c]=0
        histOrg[c]+=1
   #     print(c,ord(line2[nn:nn+1]),line2[nn:nn+1],type(c),type(line2[nn:nn+1]))
        if c != ord(line2[nn:nn+1]):
             print(mm,nn,c,line2[nn:nn+1],ord(line2[nn:nn+1]),c ==ord(line2[nn:nn+1]))
              
  
```

```python

def getAsciiCodes():
    codesdf = pd.read_excel("/home/joe/Python/ascii-codes.xlsx",engine='openpyxl')
    asciiCodes={}
    for index,row in codesdf.iterrows():
        asciiCodes[row['DEC']] = {}
        asciiCodes[row['DEC']]['symbol'] = row['Symbol']
        asciiCodes[row['DEC']]['desc'] = row['Description']
    return asciiCodes

asciiCodes=getAsciiCodes()
for val,count in sorted(histTrn.items()):
    print(val,count,asciiCodes[val]['desc'])
```
