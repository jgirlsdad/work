```python
def missingAnal(df):
    nrows = df.shape[0]
    print("Total Rows: ",nrows)
    print(f"{'Column':30.30s}   missing   %-miss   Type")
    print(f"-------------------------------------------------------------")
    for col in sorted(df.columns):
        nc = df[col].isna().sum()
        pp = df[col].isna().sum()/nrows*100
        print(f"{col:30.30s}  {nc:8d}   %{pp:4.1f}   {df[col].dtype}")
```

```python
def invYrMo(df,colDate,formt,type=0):
    if type == 0:  # simple, quick format 
        a = df.copy()
        a["jlcDateInv"] = pd.to_datetime(df[colDate],format=formt)
        a["jlcYear"] = a["jlcDateInv"].dt.year
        a["jlcMonth"] = a["jlcDateInv"].dt.month
        inv = a.groupby(["jlcYear","jlcMonth"]).count()    
    elif type == 1:  # year-month-day, with year and month as columns and thne each day as column w/ a count
        stats = {}
        a = df.copy()
        a["jlcDateInv"] = pd.to_datetime(df[colDate],format=formt)
        a["jlcYear"] = a["jlcDateInv"].dt.year
        a["jlcMonth"] = a["jlcDateInv"].dt.month
        a["jlcDay"] = a["jlcDateInv"].dt.day
        b = a.groupby(["jlcYear","jlcMonth","jlcDay"],as_index=False).count()
        for yr in range(b["jlcYear"].min(),b["jlcYear"].max()+1):
            stats[yr] = {}
            monthLast = 12
            if yr == pd.Timestamp.today().year:
                monthLast = pd.Timestamp.today().month
            for mo in range(1,monthLast+1):
               stats[yr][mo] = {}
               for dy in range(1,32):
                stats[yr][mo][dy]=0

        for indx,row in b.iterrows():
            yr = row["jlcYear"]
            mo = row["jlcMonth"]
            dy = row["jlcDay"]
            cnt = row[colDate]
            stats[yr][mo][dy]=cnt

        nrec=0
        for yr in range(b["jlcYear"].min(),b["jlcYear"].max()+1):
             monthLast = 12
             if yr == pd.Timestamp.today().year:
                monthLast = pd.Timestamp.today().month
             for mo in range(1,monthLast+1):
                lu = {f"Day {nn}":[stats[yr][mo][nn]] for nn in range(1,32)}
                l = {"year":[yr],"month":[mo]}
                l.update(lu)
                tmp = pd.DataFrame(l)
                nrec+=1
                if nrec > 1:
                   inv = pd.concat([inv,tmp])
                else:
                   inv = tmp
    
    return inv

# inv = invYrMo(dfBusCIM,"entityformdate",'%m/%d/%Y',1)
# display(inv.head())
# display(inv.tail())
```
