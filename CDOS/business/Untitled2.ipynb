```python
fin2 = open("/home/joe/bic_etl/cdos/business/business/data_source/trademarks.tsv")
linesTM = fin2.readlines()
fin2.close()
```

```python
def  analyzeString(string,hist,type=1): 
    for c in string:
        if type == 1:
            nc=ord(c)
        else:
            nc=c
        if nc not in hist:
            hist[nc]=0
        hist[nc]+=1
    return hist

hist={}
histNFields={}
histNDoubleQuotes={}
for nline,line in enumerate(linesTM):
    hist=analyzeString(line,hist,1)
    spl=line.split("\t")
    nn=len(spl)
    if nn not in histNFields:
        histNFields[nn]=0
    histNFields[nn]+=1     
    for val in spl:
        mm=val.count('"')
        if mm not in histNDoubleQuotes:
            histNDoubleQuotes[mm]=0
        histNDoubleQuotes[mm]+=1

    # for val in spl:
    #     if val.count('"') != 2 and val.count('"') != 0: 
    #         print(val.count('"'),val)
```
