```python
import pandas as pd
import numpy as np

import warnings

from pandas.core.common import SettingWithCopyWarning

warnings.simplefilter(action="ignore", category=SettingWithCopyWarning)
```

```python
dfBusCIM = pd.read_csv("Business_Entities_in_Colorado.csv")
```

```python
dfXen = pd.read_csv("~/bic_etl/cdos/business/business/data_transformed/business_entities.csv")
```

```python
def printComps(eid,how):
#    eid = 20011097520
    a = dfBusCIM.loc[dfBusCIM["entityid"] == eid]
    b = dfXen.loc[dfXen["entityId"] == eid]
    a.drop("entityformdate",inplace=True,axis=1)
    b.drop("entityFormDate",inplace=True,axis=1)

#     a = dfA.copy()
#     b = dfB.copy()
   
    cols=[]
    org = []
    new = []
    for col in b:  
        if isinstance(a[col.lower()].values[0],str):
            a[col.lower()] = a[col.lower()].values[0].strip()
        if isinstance(b[col].values[0],str):
            b[col].values[0] = b[col].values[0].strip()
            
        if how == 0: 
            cols.append(col)
            org.append(a[col.lower()].values[0])
            new.append(b[col].values[0])
        else:
            if col.lower() == "entityid":
                cols.append(col)
                org.append(a[col.lower()].values[0])
                new.append(b[col].values[0])
            
            if type(a[col.lower()].values[0]) != type(b[col].values[0]):
          #       print(type(a[col.lower()].values[0]),type(b[col].values[0]),a[col.lower()].values[0],b[col].values[0])
                 cols.append(col)
                 org.append(a[col.lower()].values[0])
                 new.append(b[col].values[0])
            elif  a[col.lower()].values[0] !=  b[col].values[0]:
                 if isinstance(a[col.lower()].values[0],str) and isinstance(b[col].values[0],str):
                     cols.append(col)
                     org.append(a[col.lower()].values[0])
                     new.append(b[col].values[0])
                 elif not np.isnan(a[col.lower()].values[0]) or  not np.isnan(b[col].values[0]):
                     cols.append(col)
                     org.append(a[col.lower()].values[0])
                     new.append(b[col].values[0])


    c = pd.DataFrame({"Column":cols,"Orig Val":org,"New Val":new})
    print(c["Orig Val"].equals(c["New Val"]))
    display(c)
    
eid = 20011097520
#printComps(dfBusCIM.loc[dfBusCIM["entityid"] == eid].drop("entityformdate",axis=1),dfXen.loc[dfXen["entityId"] == eid].drop("entityFormDate",axis=1),0)
printComps(20011097520,1)
```
