```python
import pandas as pd
tdir="/home/joe/bic_etl/cdos/lobbyist/data_source/"
# file=input("Input File")
ffiles = open("listFiles","r")
files=ffiles.readlines()


dfs={}
for file in files:
    file=file.strip()
    file=file.replace("txt","tsv")
    print(file)
    try:
       dfs[file]=pd.read_csv(f"{tdir}{file}",delimiter="\t")
    except:
       dfs[file]=pd.read_csv(f"{tdir}{file}",delimiter="\t",encoding="latin")
        
```

```python
xrefs = {'prof_bills.tsv':  "ProfessionalLobbyistBills-fields.tsv",
'prof_clients.tsv':"ProfessionalLobbyistClients-fields.tsv",
'prof_disclosure_report_summary.tsv':"ProfessionalLobbyistDisclosur-fields.tsv",
'prof_expenditures.tsv':"ProfessionalLobbyistExpenditu-fields.tsv",
'prof_income.tsv':"ProfessionalLobbyistIncome-fields.tsv",
'prof_lobbyist_directory.tsv':"ProfessionalLobbyistDirectory-fields.tsv",
'prof_subcontractors.tsv':"ProfessionalLobbyistSubcontra-fields.tsv",
'state_lobbyist_bills.tsv':"StateLobbyistBills-fields.tsv",
'state_lobbyist_directory.tsv':"StateLobbyistDirectory-fields.tsv",
'state_lobbyist_funds_hours.tsv':"StateLobbyistsFundsandHours-fields.tsv",
'state_lobbyist_officials.tsv':"StateLobbyistOfficials-fields.tsv"
}
```

```python
bad={}
dctsMet={}
for file in dfs:
    infile=f"defs/{xrefs[file]}"
    fin = open(infile,"r")
    lines = fin.readlines()

    fin.close()
    fields = {}
    print(file,infile)
    for line in lines:
        spl = line.split("\t")
        if len(spl) != 2:
            print(spl)
        fields[spl[0]] = spl[1].strip().strip("\n")
    colDat = set(list(dfs[file].columns))
    colMet = set(list(fields.keys()))
    dctsMet[file] = list(fields.keys())
  
    
    
    a = colDat-colMet
    b = colMet-colDat
    # if file == 'prof_subcontractors.tsv':
    #     print("L! ",len(colDat))
    #     for val in fields.keys():
    #         print(val)
    #     print("LL ",len(colMet))
    if len(a) > 0:
        print(file,"Not Found in Metadata",a)
        bad[file]={}
        bad[file]['Not in Meta']=a
    if len(b) > 0:
        print(file,"Not Found in Data",b)
        if file not in bad:
            bad[file]={}
        bad[file]['Not in Data'] = b
    
    print("-------------------------------------")

                 
        
```

```python
for file,dct in bad.items():
    print(file)
    for k,lst in dct.items():
        print('  ',k)
        for fld in lst:
            print("    ",fld)
    print("---")
```
