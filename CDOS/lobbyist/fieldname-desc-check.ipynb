# Check Field Names and Descriptions

```python
import pandas as pd
from sodapy import Socrata
import pandas as pd
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', None)
```

```python
## Get datasets on CIM
cim_url_query = 'data.colorado.gov'
datasets = None

with Socrata(cim_url_query, None) as client:
    datasets = client.datasets()
```

```python
len(datasets)
```

```python
w4x4s = []
dfs = {}
for dataset in datasets:
    if "lobbyist" in dataset["resource"]["name"].lower():
            title = dataset['resource']['name']
            w4x4 = dataset['resource']['id']
            print(w4x4,title)
            w4x4s.append(w4x4)
```

```python
w4x4s.pop(0)
w4x4s
```

```python
badTitles=[]
badCols = []
badFields = []
badDesc = []
bad4x4s = []
dfs = {}
datasetLobby = []
for dataset in datasets:
    if "lobbyist" in dataset["resource"]["name"].lower():
        if dataset["resource"]["id"] in w4x4s:
            title = dataset['resource']['name']
            w4x4 = dataset['resource']['id']
            datasetLobby.append(dataset)
            cols=dataset['resource']['columns_name']
            fields=dataset['resource']['columns_field_name']     
            dtype=dataset['resource']['columns_datatype']  
            descr=dataset['resource']['columns_description']   
            format=dataset['resource']['columns_format']  
            
            df = pd.DataFrame({
                    "Cols":cols,
                    "Fields":fields,
                    "Description":descr,
                    "Date Type":dtype})
            dfs[title]=df
            for nn,col in enumerate(cols):
                field=fields[nn]
                desc=descr[nn]    
                if len(col) < 1 or len(field) < 1 or len(desc) < 1:
                    print(f"{title:.30s} {w4x4:7s}  {col:.30s}  {field:30.30s} {desc}")
                    badTitles.append(title)
                    bad4x4s.append(w4x4)
                    badCols.append(col)
                    badFields.append(field)
                    badDesc.append(desc)
```

```python
dfBad = pd.DataFrame({
    "Title":badTitles,
    "4x4":bad4x4s,
    "Col":badCols,
    "Field":badFields,
    "Description":badDesc
    })
```
