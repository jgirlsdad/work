```python
import os,sys,pandas as pd,difflib
bic_etl_home = os.environ['bic_etl_home']
sys.path.insert(0, os.path.join(bic_etl_home, 'general', 'scripts'))
sys.path.insert(0, '/home/joe/work/myLibs')
from datetime import datetime, date
import json
from sodapy import Socrata

# Get today's date as a date object
today = date.today()
today_str = today.strftime("%Y-%m-%d")
import funcs
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)  # Show all columns
pd.set_option('display.width', None)        # Don't wrap output
pd.set_option('display.max_colwidth', None) # Show full column content

```

## Source Data File to X-Reference Dictionary

```python
title = 'Current Notaries in Colorado'
srcFile = "/home/joe/bic_etl/cdos/government/data_source/CurrentCommissionedNotaries.tsv"
sep = '\t'
def createSrcTrnLib(title,srcFile,outFile,inSep=''):
    cim_url_query = "data.colorado.gov"
    cimDatasets = {}
    with Socrata(cim_url_query, None) as client:
        datasets = client.datasets()
        for dataset in datasets:
            if dataset['owner']['display_name'] == 'Business Intelligence Center of CO':
                titleC=dataset["resource"]["name"]
                cimDatasets[titleC]=dataset
                if title == titleC:
                    break

    desc=cimDatasets[title]['resource']['columns_description']
    cimCols=cimDatasets[title]['resource']['columns_name']
    w4x4=cimDatasets[title]['resource']['id']
    if len(inSep) > 0:
        sep=inSep
    else:
        sep=","
    df=pd.read_csv(srcFile, sep=sep)
    cols = list(df.columns)
    lib = {w4x4:{}}
    for nn,col in enumerate(cimCols):
        col_desc = desc[nn]
        best_match = difflib.get_close_matches(col, cols, n=1, cutoff=0.6)[0]

        lib[w4x4][best_match]={'xref':col,'desc':col_desc,'notes':'','date':today_str}

    with open(outFile, 'w') as f:
        json.dump(lib, f, indent=2)
    print("DONE")

createSrcTrnLib(title,srcFile,outFile='test2.json',sep)
```

## CIM Fields to Source X-Reference Dictionary

```python
title = "Septic Systems in Boulder County Colorado"
fileSource = "/home/joe/bic_etl/boulder/data_source/septic_export.csv"
cim_url_query = "data.colorado.gov"
cimDatasets = {}
with Socrata(cim_url_query, None) as client:
    datasets = client.datasets()
    for dataset in datasets:
        if dataset['owner']['display_name'] == 'Business Intelligence Center of CO':
            titleC=dataset["resource"]["name"]
            cimDatasets[titleC]=dataset
            if title == titleC:
                break

desc=cimDatasets[title]['resource']['columns_description']
cimCols=cimDatasets[title]['resource']['columns_name']
w4x4=cimDatasets[title]['resource']['id']
df=pd.read_csv(fileSource, sep=",", encoding='latin')
cols = list(df.columns)
lib = {w4x4:{}}
none=0
for nn,col in enumerate(cimCols):
    col_desc = desc[nn]
    best_match = difflib.get_close_matches(col, cols, n=1, cutoff=0.6)
    if len(best_match) > 0:
        best_match = best_match[0] 
    else:
        none+=1
        best_match = f"none-{none}"
        print(f"No match for cim col: {col}")
    lib[w4x4][best_match]={'xref':col,'desc':col_desc,'notes':'','date':today_str}

with open('test.json', 'w') as f:
    json.dump(lib, f, indent=2)
print("DONE")
```
