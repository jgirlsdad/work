```python
fin = open("list.jup","r")
jup = {}
for line in fin:
    spl = line.split()
    jup[spl[0]] = spl[1]
    
    
fin = open("list.sys","r")
syt = {}
for line in fin:
    spl = line.split()
    syt[spl[0]] = spl[1]
```

```python
nmatchs=0
for ky,val in syt.items():
    if ky in jup:
        if jup[ky] != syt[ky]:
            print(ky,jup[ky],syt[ky])
        else:
            nmatchs+=1
    else:
        print("Not Found ",ky)
        
print(nmatchs)
```
