```python
df["Date"] = pd.to_datetime(df["Timestamp"],format="%m/%d/%Y %I:%M:%S %p")
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
```

```python
hist={}
for index,row in df.iterrows():
    w4x4=row["Asset UID"]
    access=row["Access Type"]    
    name=row["Asset Name"]  
    assetType=row["Asset Type"]     
    year=row["Year"]        
    month=row["Month"]   
    
    if w4x4 not in hist:
        hist[w4x4]={}
    if access not in hist[w4x4]:
        hist[w4x4][access]={}
    if year not in hist[w4x4][access]:
        hist[w4x4][access][year]={}
    if month not in hist[w4x4][access][year]:
        hist[w4x4][access][year][month]= 0
        
    hist[w4x4][access][year][month]+=1    
```

```python
pd.DataFrame.from_dict(hist)
```

```python
#hist[w4x4][access][year][month]+=1    
datasets = []
access= []
years = []
months = []
counts = []

for w4x4,v1 in hist.items():
     for acc,v2 in v1.items():
         for year,v3 in v2.items():
            for month,count in v3.items():
                datasets.append(w4x4)
                access.append(acc)
                years.append(year)
                months.append(month)
                counts.append(count)
        
```

```python
stats=pd.DataFrame({
    "4x4":datasets,
    "access":access,
    "year":years,
    "month":months,
    "count":counts
})
```
