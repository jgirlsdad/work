```python
import fiona
from collections import OrderedDict

GDB_PATH = r"GRDMS_Pub_CIM.gdb/GRDMS_Pub_CIM.gdb"

def inspect_gdb(gdb_path):
    print(f"\nInspecting GDB: {gdb_path}\n")

    layers = fiona.listlayers(gdb_path)

    summary = []

    for layer in layers:
        with fiona.open(gdb_path, layer=layer) as src:
            schema = src.schema
            fields = list(schema["properties"].keys())
            geom = schema["geometry"]
            row_count = len(src)

            summary.append(OrderedDict({
                "dataset": layer,
                "rows": row_count,
                "fields": len(fields),
                "geometry": geom,
                "field_names": fields
            }))

            print(f"Dataset: {layer}")
            print(f"  Rows: {row_count}")
            print(f"  Fields: {len(fields)}")
            print(f"  Geometry: {geom}")
            print("")

    return summary


if __name__ == "__main__":
    inspect_gdb(GDB_PATH)
```

```python
import fiona
import geopandas as gpd
import os

GDB_PATH = r"GRDMS_Pub_CIM.gdb/GRDMS_Pub_CIM.gdb"
OUT_DIR = r"./csv"

os.makedirs(OUT_DIR, exist_ok=True)

layers = fiona.listlayers(GDB_PATH)

print(f"Found {len(layers)} layers\n")

for layer in layers:
    print(f"Processing layer: {layer}")

    gdf = gpd.read_file(GDB_PATH, layer=layer)

    # Drop geometry if present
    #df = gdf.drop(columns="geometry", errors="ignore")

    out_csv = os.path.join(OUT_DIR, f"{layer}.csv")
    gdf.to_csv(out_csv, index=False)

    print(f"  → wrote {out_csv}")

print("\nAll layers exported.")
```
